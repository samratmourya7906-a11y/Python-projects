```python
# QUE.1
numbers = [20, 55, 30, 80, 45, 90, 10]

result = [num for num in numbers if num > 50]

print(result)
```

```python
# QUE.2
numbers = [10, 25, 35, 60, 45, 15, 50, 70]

result = [num for num in numbers if 20 >= num <= 50]

print( result)
```

```python
# QUE.3
numbers = [10, 15, 20, 30, 45, 50, 60, 75]

result = [num for num in numbers if num % 3 == 0 and num % 5 == 0]

print( result)
```

```python
# QUE.4
names = ["Amit", "Rahul", "Neha", "Rohit", "Sneha"]

result = [name for name in names if len(name) > 4]

print(result)
```

```python
# QUE.5
names = ["Amit", "Rahul", "Ankit", "Neha", "Arjun", "Rohit"]

result = [name for name in names if name.startswith("A")]

print(result)
```

```python
# QUE.6
text = "programming"

result = [char for char in text if char in "aeiou"]

print( result)
```

```python
# QUE.7
data = "Python123Programming456"

result = [char for char in data if char.isdigit()]

print( result)
```

```python
# QUE.8
numbers = [10, 15, 20, 25, 30]

result = ["Even" if num % 2 == 0 else "Odd" for num in numbers]

print( result)
```

```python
# QUE.9
numbers = [10, -5, 20, -8, 30, -2]

result = [num if num >= 0 else 0 for num in numbers]

print( result)
```

```python
# QUE.10
prices = [500, 1200, 800, 2000, 1500]

result = [price * 0.90 if price >= 1000 else price * 0.95 for price in prices]

print( result)
```

```python
# QUE.11
employees = [
    {"name": "Amit", "salary": 45000},
    {"name": "Rahul", "salary": 60000},
    {"name": "Neha", "salary": 75000},
    {"name": "Rohit", "salary": 35000}
]

result = [employee["name"] for employee in employees if employee["salary"] > 50000]

print( result)
```

```python
#Q.15
employees = [
    {"name": "Amit", "department": "IT"},
    {"name": "Rahul", "department": "HR"},
    {"name": "Neha", "department": "IT"},
    {"name": "Rohit", "department": "Sales"}
]
result = [ex['name'] for ex in employees if ex['department']=="IT"]
print(result)
```
